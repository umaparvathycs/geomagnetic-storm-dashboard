# 03 Catalogue analysis and scientific validation
Inspect known storms against the Kyoto monthly pages before trusting the catalogue.

```python
import sys, pathlib
ROOT = pathlib.Path.cwd().parent if pathlib.Path.cwd().name == 'notebooks' else pathlib.Path.cwd()
sys.path.insert(0, str(ROOT))
import pandas as pd, numpy as np, json
import matplotlib.pyplot as plt
from src.common import load_config, resolve
cfg = load_config(ROOT / 'config.yaml')
```

```python
cat = pd.read_csv(resolve(cfg,'catalogue'), parse_dates=['start_time','peak_time','end_time'])
print(cat.intensity_tier.value_counts())
print(cat[['total_duration_hours','main_phase_hours','recovery_phase_hours','descent_rate_nt_hr','recovery_to_150_hours']].describe())
print(cat.data_quality_flag.value_counts())
```

```python
cat.groupby(cat.start_time.dt.year).size().plot(kind='bar', figsize=(14,3), title='Storms per year'); plt.show()
```
